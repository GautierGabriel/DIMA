```python
import json
import matplotlib.pyplot as plt

def plot_learning_curves(json_path, output_name='loss_curve.pdf'):
    # 1. Chargement des données
    with open(json_path, 'r') as f:
        data = json.load(f)
    
    # Extraction et modification : on ajoute +1 à la validation
    train_loss = data.get('train_loss', [])
    val_loss = [v for v in data.get('val_loss', [])]
    
    epochs = range(1, len(train_loss) + 1)

    # 2. Configuration de la figure
    fig, ax = plt.subplots(figsize=(7, 4.5))

    # 3. Tracé des courbes
    ax.plot(epochs, train_loss, color='#1f77b4', linewidth=1.5, label='Train Loss')
    ax.plot(epochs, val_loss, color='#9467bd', linewidth=1.5, label='Validation Loss')

    # 4. Esthétique
    ax.set_xlabel('Epochs', fontsize=12)
    ax.set_ylabel('Loss', fontsize=12)
    ax.set_title('Model Training Dynamics', fontsize=14, pad=15)

    ax.grid(True, linestyle=':', alpha=0.6)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    ax.spines['left'].set_linewidth(1.2)
    ax.spines['bottom'].set_linewidth(1.2)

    ax.legend(frameon=False, fontsize=11, loc='upper right')

    # 5. Sauvegarde
    plt.tight_layout()
    plt.savefig(output_name, dpi=300, bbox_inches='tight') 
    print(f"✅ Courbe sauvegardée sous : {output_name}")
    plt.show()
```

```python

plot_learning_curves('loss_nuc.json')
```
