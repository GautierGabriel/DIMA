```python
import sys
import numpy as np
sys.path.append('/content/scripts')

import visualisation 
visualizer = visualisation.CellVisualizer()

mask = visualizer.load_data("001", mode = 'Cell_Pred')
visualizer.plot('001', mode = 'Cell_Pred')

_, counts = np.unique(mask[mask > 0], return_counts=True)

print(f"Nombre de cellules : {len(counts)}")
print(f"Surface totale     : {np.sum(counts) if len(counts) > 0 else 0}")
print(f"Surface moyenne    : {np.mean(counts) if len(counts) > 0 else 0:.2f}")
```
