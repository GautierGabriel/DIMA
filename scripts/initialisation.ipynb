```python
import owncloud
import getpass
import sys
sys.path.append('/content/scripts')

url = 'https://cloud.minesparis.psl.eu' 
user = 'gabriel.gautier'

password = getpass.getpass(f"Entrez le mot de passe ownCloud pour {user} : ")

oc = owncloud.Client(url)
oc.login(user, password)

print("Connecté à ownCloud Mines")
```

```python
import shutil
from datetime import datetime
import os

print("Préparation de la synchronisation...")

date_str = datetime.now().strftime("%Y-%m-%d_%Hh%M")
nom_archive = f"resultats_{date_str}"
nom_zip = f"{nom_archive}.zip"

shutil.make_archive(nom_archive, 'zip', './data/processed/')

# --- Création du dossier distant s'il n'existe pas ---
dossier_distant = '/travail/Mines/DIMA/Segmentation/data/processed/'
try:
    oc.mkdir(dossier_distant)
except:
    pass

remote_path = f"{dossier_distant}{nom_zip}"
print(f"Envoi de {nom_zip} vers ownCloud...")
oc.put_file(remote_path, nom_zip)

!unzip -o -q {nom_zip} -d .

os.remove(nom_zip)

print("Synchronisé avec ownCloud Mines")
```
