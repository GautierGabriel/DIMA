```python
import matplotlib.pyplot as plt
import cv2
from pathlib import Path

def plot (i):

    dossier = Path('../data/1-Images/01-Data/')
    fichier = next(dossier.glob(i +'*'))

    img = cv2.imread(str(fichier), cv2.IMREAD_UNCHANGED)

    plt.imshow(img, cmap='gray')
    plt.title(f"Image : {fichier.name}")
    plt.show()
```

```python
plot('045')
```
